# Merging Data -  pd.merge()function

Usually, the data come from multiple tables or files. So, before doing any data analysis, one needs to combine the files.

We will use the pandas library for that, so, first,

import the pandas library with the standard alias 'pd'


```python
import pandas as pd
```

Let's create some tables:

```python
df1 = pd.DataFrame({
    "state": ["A", "B", "C", "K"],
    "x": [1, 2, 4, 5]
})

print(df1)

df2 = pd.DataFrame({
    "state": ["A", "B", "C", "D"],
    "y": [10, 20, 40, 50]
})

print(df2)
```



#  Inner join

returns rows when there is a match in both tables.

`inner_join = pd.merge(df1, df2, on="state", how="inner")`

* **pd.merge()** is the function from the
pandas library that performs the merge operation.
* **df1** and **df2** are the two DataFrames we want to merge.
* **on="state"** specifies the column to use for joining the two DataFrames. The merge function will look for matching values in the "state" column in both df1 and df2.
* **how="inner"** specifies the type of join. An "inner" join returns only the rows where the join column ("state" in this case) has matching values in both DataFrames. Rows with no match in either DataFrame are excluded.

```python
inner_join = pd.merge(df1, df2, on="state", how="inner")
print(inner_join)
```

# Left join

returns all rows from the left table, even if there are no matches in the right table.


```python
left_join = pd.merge(df1, df2, on="state", how="left")
print(left_join)
```

# Right join
returns all rows from the right table, even if there are no matches in the left table.

```python
right_join = pd.merge(df1, df2, on="state", how="right")
print(right_join)
```

# Full join
returns all rows from both the tables and fill in NaN for missing matches on either side.

```python
full_join = pd.merge(df1, df2, on="state", how="outer")
print(full_join)
```

# What if column names don't match?

```python
df3 = pd.DataFrame({
    "id": ["A", "B", "C", "D"],
    "y": [10, 20, 40, 50]
})

print(df3)
print(df1)
```

we can specify col names to perform the merge:

`merged_df = pd.merge(df1, df3, left_on="state", right_on="id", how="left")`

* pd.merge(df1, df3, ...) indicates that we are merging df1 and df3.
* left_on="state" tells pandas to use the "state" column from the left DataFrame (df1) as the merge key/variable.
* right_on="id" tells pandas to use the "id" column from the right DataFrame (df3) as the merge key/variable.
* how="left" specifies the type of join (you can use "inner", "right", or "outer" here as well, depending on your needs).



```python
merged_df = pd.merge(df1, df3, left_on="state", right_on="id", how="left")
print(merged_df)
```

```python
merged_df13 = pd.merge(df1, df3, left_on="state", right_on="id", how="inner")
print(merged_df13)
```

* how="left": We're performing an left join here. You can change this to "inner", "right", or "outer" depending on your needs.

# Where did the rows in your merged dataframe come from?

Adding **the indicator** parameter is very useful for understanding the results of your merge operation, as well as for filtering or analisys of the rows based on whether they had a match in one or both of the original DataFrames.


```python
merged_with_indicator = pd.merge(df1, df2, on="state", how="outer", indicator=True)
print(merged_with_indicator)
```

Then you can tabulate the _merge variable to quicky check the quality of your merge

```python
merged_with_indicator['_merge'].value_counts()
```

# Appending data - pd.concat()

```python
dfa1 = pd.DataFrame({
    "Age": [21, 15],
    "Name": ["John", "April"]
})

dfa2 = pd.DataFrame({
    "Name": ["Oscar", "Polly"],
    "Age": [45, 50]
})

print(dfa1)
print(dfa2)
```

Here is how to use the pd.concat() function:

`dfa1_dfa2 = pd.concat([dfa1, dfa2], ignore_index=True)`

* pd.concat([dfa1, dfa2]) combines (concatenates) the two DataFrames vertically.
* `ignore_index=True` resets the index of the resulting DataFrame, which is often useful when appending.

```python
dfa1_dfa2 = pd.concat([dfa1, dfa2], ignore_index=True)

print(dfa1_dfa2)
```

```python
pd.concat([dfa1, dfa2])
```

What if variables are not the same in the two dataframes?
Not a probelm at all!

```python
dfa3 = pd.DataFrame({
    "Name": ["James", "Emily"],
    "State": ["PA", "FL"]
})

print(dfa3)
print(dfa1)
```

```python
dfa1_dfa3 = pd.concat([dfa1, dfa3], ignore_index=True)

print(dfa1_dfa3)
```

Note: there is no `indicator` option for the `pd.concat()` function.
If you want to keep track of the data source,
you need to manually add a variable (e.g., "source") to the original DataFrames BEFORE appending them.

```python
dfa1['source'] = 'dfa1'
dfa3['source'] = 'dfa3'
print(dfa1)
print(dfa3)
dfa1_dfa3_ind = pd.concat(
    [dfa1, dfa3])

print(dfa1_dfa3_ind)
```

# Pivoting - pd.melt()

```python
data = pd.DataFrame({
    'country': ['Afghanistan', 'Brazil', 'China'],
    '1999': [19987071, 172006362, 1272915272],
    '2000': [20595360, 174504898, 1280428583]
})

print(data)
```

```python
df_long = pd.melt(data,
  # Columns to keep as identifier variables:
                  id_vars=['country'],
  # Columns to stack into one:
                   value_vars=['1999', '2000'],
  # Name for the stacked column:
                  var_name='year',
  # Name for the new column holding the unpivoted values
                  value_name='population')


print(df_long)
```

*Explanation:*

* **`pd.melt(data, ...)`**: Applies the melt function to the "data" DataFrame.
* **`id_vars=['country']`**: Specifies the column(s) that should remain as identifier variables. In this case, 'country' remains a distinct column.
* **`value_vars=['1999', '2000']`**: Specifies the columns that you want to stack (unpivot or melt). The data from these columns will be stacked into a single column. **NOTE** pd.melt() automatically treats all columns not listed in **`id_vars`** as the columns to stack/unpivot - so you do not need to specify them!
* **`var_name='year'`**: This is the name you want for the new column that will contain the names of the columns that were unpivoted (i.e., '1999' and '2000').
* **`value_name='population'`**: This is the name you want for the new column that will contain the values from the unpivoted columns.


As a result we get a DataFrame with three columns:

`'country', 'year', and 'population', `

achieving the same transformation we did in R using the pivot_longer().
