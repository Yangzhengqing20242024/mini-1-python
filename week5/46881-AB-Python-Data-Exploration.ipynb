By adding it in this case, we will sort the original DataFrame.

# Filtering

Let's find games that happened on Saturdays and Sundays.

First, let's see if we have Saturdays or Sundays games using


***.isin() method**

# Recall %in% in R

The .isin() function returns a Series of the same size and shape, as the original but with True and False in place of all values, depending on whether the value from the original Series was included in the list passed, i.e.
whetehr the games was on Saturdays or Sundays.


```python
dodgers["day_of_week"].isin(["Saturday", "Sunday"])
```

Use the tilde operator at the beginning of the code block to invert the True/ False values so that the function returns True if the value is NOT in the list passed

```python
~dodgers["day_of_week"].isin(["Saturday", "Sunday"])
```

Let's keep only the Saturdays or Sundays games. Let's create a new dataframe, call it "dodgersSS"

```python
dodgersSS=dodgers[dodgers["day_of_week"].isin(["Saturday", "Sunday"])]
print(dodgersSS.info())
print(dodgersSS.head(10))
```

# Summary by groups

groupby() groups data by category

use aggregating function to compute summary statistics by groups

`dodgers.groupby("day_of_week").mean()` will return an error. ***Why?***



*Ans:* The groupby() operation calculates the mean for all applicable columns within each group. But the dodgers DataFrame contains several categorical variables. One cannot compute the means for all variables!

What do you think this will return?

`dodgers.groupby("day_of_week").mean("attend")`

Run it!

```python
dodgers.groupby("day_of_week").mean("attend")
```

The mean aggregation applied to all numerical columns...

To find the mean  ONLY for the 'attend' column for each group:

Group by 'day_of_week' and then select the 'attend' column to calculate its mean



```python
dodgers.groupby("day_of_week")["attend"].mean()
```

```python
dodgers_attend_mean_by_day = dodgers.groupby("day_of_week")["attend"].mean()
print(dodgers_attend_mean_by_day)
```

**Other aggregating functions:


*   Sum: .sum()
*   Minumum: .min()
*   Maximum: .max()
*   Count: .count()
*   Variance: .var()
*   Standard deviation: .std()
*   Median: .median()



```python
dodgers.groupby("day_of_week")["attend"].median()
```

** .agg() **

If we want to compute more than one sum stats at once, we can use .agg() method,

passing a list of statistics we want to compute.

```python
dodgers["attend"].agg(["mean","median"])
```

```python
dodgers[["attend","temp"]].agg(["mean","median"])
```

Now let's compute sum stats for the attendance by the day of the week variable:

```python
dodgers.groupby("day_of_week")["attend"].agg(["mean","median"])
```

```python
dodgers.groupby("day_of_week")[["attend", "temp"]].agg(["mean","median"])
```

```python
dodgers.groupby("day_of_week")[["attend", "temp"]].agg("mean")
```

We can create names for the columns...by creating named tuples inside the .agg() function.

Each named tuple should include a column name followed by the aggregating function to apply to that column. The name of the tuple will be the name of the resulting column.

```python
dodgers.groupby("day_of_week")[["attend", "temp"]].agg(
    mean_attend=("attend", "mean"),
    mean_temp=("temp", "mean")
)
```

```python
sum_stats1=dodgers.groupby("day_of_week")[["attend", "temp"]].agg(
    mean_attend=("attend", "mean"),
    mean_temp=("temp", "mean")
)
print(sum_stats1)
```

```python
dodgers.groupby("day_of_week")["attend"].agg(
    mean_attend="mean",
    median_attend="median"
)
```

```python
sns.barplot(data=dodgers, x="day_of_week", y="attend")
plt.show()
```

Note that .agg() function collapses the data by group, giving you summary statistics for each group. So you end up with a smaller result table than your original data.

Use **.transform()**to perform a group-specific calculation and return a result that has the same index as the original DataFrame. We will look at the example of .transform() later.

# Missing data

Let's create a demo dataset:

```python
demo_data = pd.DataFrame({
    'year': [2021, 2023, 2023, 2024, 2025],
    'population': [100, 200, None, None, 500]
})
print(demo_data)
demo_data.head(5)
```

How we can check if we have any missing data?

```python
demo_data.info()
```

Another way is to use the .isna() function

```python
demo_data.isna()
```

```python
~demo_data.isna()
```

We can count the number of missing values per column/variable by **chaining** the .isna() and .sum() methods:

```python
demo_data.isna().sum()
```

As alwasy, one way to deal with missing data is just to drop the observations with missing data....

But suppose you want to replace them with some values instead,

i.e. some summary statistics.

Let's replace the missing values with the mean.
To do that we wil use ** .fillna()** method


```python
demo_data["population"].fillna(demo_data["population"].mean())
```

# Add a new column

Let's add a new column 'weekend_game' indicating games happend on Saturday or Sunday

Steps:

1. Use .str.contains "Saturday|Sunday" to find weekend days
2. Convert the boolean result (True/False) to integers (1/0) using .astype(int)

```python
dodgers['weekend_game'] = (
    dodgers["day_of_week"]
     .str.contains("Saturday|Sunday")
)
print(dodgers.head())

dodgers['is_weekend'] = (
    dodgers["day_of_week"]
    .str.contains("Saturday|Sunday")
    .astype(int)
)
```
